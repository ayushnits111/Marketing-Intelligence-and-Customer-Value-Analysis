```python
import pandas as pd
import mysql.connector

# Connect to MySQL
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Amisha@2004",
    database="marketing_db"
)

# Create customer-level dataset
query = """
SELECT
    C.customer_id,
    C.acquisition_channel,
    C.acquisition_cost,
    COUNT(T.transaction_id) AS transaction_count,
    COALESCE(SUM(T.revenue), 0) AS total_revenue,
    COALESCE(AVG(T.revenue), 0) AS average_revenue
FROM Customers C
LEFT JOIN Transactions T
    ON C.customer_id = T.customer_id
GROUP BY
    C.customer_id,
    C.acquisition_channel,
    C.acquisition_cost
"""

df = pd.read_sql(query, conn)

conn.close()

# Display the dataset
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)
```

```python
# Create customer-level Profit
df["profit"] = df["total_revenue"] - df["acquisition_cost"]

# Create customer-level ROI
df["roi"] = (
    (df["total_revenue"] - df["acquisition_cost"])
    / df["acquisition_cost"]
) * 100

# Create conversion indicator
df["conversion"] = (df["transaction_count"] > 0).astype(int)

# Check the result
print(df.head())

print("\nUpdated Dataset Shape:")
print(df.shape)

print("\nSummary Statistics:")
print(df[[
    "acquisition_cost",
    "transaction_count",
    "total_revenue",
    "average_revenue",
    "profit",
    "roi"
]].describe())
```

```python
##Customer Segmentation using K-Means
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Select features for customer segmentation
features = df[
    [
        "acquisition_cost",
        "transaction_count",
        "total_revenue",
        "profit",
        "roi"
    ]
]

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(features)

# Create K-Means model
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

# Assign customers to clusters
df["customer_segment"] = kmeans.fit_predict(X_scaled)

# Check number of customers in each segment
print("Customers in each segment:")
print(df["customer_segment"].value_counts().sort_index())
```

```python
##Understand What Each Segment Represents
# Analyze each customer segment
segment_analysis = df.groupby("customer_segment").agg(
    customer_count=("customer_id", "count"),
    average_acquisition_cost=("acquisition_cost", "mean"),
    average_transactions=("transaction_count", "mean"),
    average_revenue=("total_revenue", "mean"),
    average_profit=("profit", "mean"),
    average_roi=("roi", "mean"),
    conversion_rate=("conversion", "mean")
).reset_index()

# Convert conversion rate to percentage
segment_analysis["conversion_rate"] *= 100

# Display the segment analysis
print(segment_analysis.round(2))
```

```python
# =====================================================
# PREPARE DATASETS FOR POWER BI
# =====================================================

# 1. Customer-level dataset with ML segments
customer_powerbi = df.copy()

# 2. Channel-level analysis
channel_powerbi = df.groupby("acquisition_channel").agg(
    customers=("customer_id", "count"),
    total_revenue=("total_revenue", "sum"),
    total_acquisition_cost=("acquisition_cost", "sum"),
    total_profit=("profit", "sum"),
    average_roi=("roi", "mean"),
    conversion_rate=("conversion", "mean")
).reset_index()

channel_powerbi["conversion_rate"] *= 100

# 3. Segment-level analysis
segment_powerbi = df.groupby("customer_segment").agg(
    customer_count=("customer_id", "count"),
    average_acquisition_cost=("acquisition_cost", "mean"),
    average_transactions=("transaction_count", "mean"),
    average_revenue=("total_revenue", "mean"),
    average_profit=("profit", "mean"),
    average_roi=("roi", "mean"),
    conversion_rate=("conversion", "mean")
).reset_index()

segment_powerbi["conversion_rate"] *= 100


# =====================================================
# EXPORT CSV FILES
# =====================================================

customer_powerbi.to_csv(
    "customer_analysis_powerbi.csv",
    index=False
)

channel_powerbi.to_csv(
    "channel_analysis_powerbi.csv",
    index=False
)

segment_powerbi.to_csv(
    "segment_analysis_powerbi.csv",
    index=False
)

print("Files created successfully!")

print("\nCustomer dataset:", customer_powerbi.shape)
print("Channel dataset:", channel_powerbi.shape)
print("Segment dataset:", segment_powerbi.shape)
```
